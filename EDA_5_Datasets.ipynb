```python
# 1. Palmer Penguins

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

url = "https://raw.githubusercontent.com/hosseinayoubi/EDA/main/penguins/penguins_size.csv"
penguins = pd.read_csv(url, na_values="NA")

print("Shape:", penguins.shape)
print("\nMissing values:")
print(penguins.isna().sum())

display(penguins.head())
display(penguins.select_dtypes(include="number").describe().round(2))

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

sns.boxplot(data=penguins, x="species", y="body_mass_g", ax=axes[0])
axes[0].set_title("Body Mass by Species")
axes[0].set_xlabel("Species")
axes[0].set_ylabel("Body Mass (g)")

sns.boxplot(data=penguins, x="species", y="flipper_length_mm", ax=axes[1])
axes[1].set_title("Flipper Length by Species")
axes[1].set_xlabel("Species")
axes[1].set_ylabel("Flipper Length (mm)")

sns.scatterplot(
    data=penguins,
    x="culmen_length_mm",
    y="culmen_depth_mm",
    hue="species",
    ax=axes[2]
)
axes[2].set_title("Bill Measurements")
axes[2].set_xlabel("Bill Length (mm)")
axes[2].set_ylabel("Bill Depth (mm)")

plt.tight_layout()
plt.show()

print("\nMy note:")
print("Gentoo penguins are clearly heavier and have longer flippers.")
print("The bill measurements also separate the species quite well.")
```

```python
# 2. Titanic

url = "https://raw.githubusercontent.com/hosseinayoubi/EDA/main/titanic/train.csv"
titanic = pd.read_csv(url)

print("Shape:", titanic.shape)
print("\nMissing values:")
print(titanic.isna().sum())

display(titanic.head())
display(titanic[["Age", "Fare", "Pclass", "Survived"]].describe().round(2))

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

sns.boxplot(data=titanic, x="Pclass", y="Age", ax=axes[0])
axes[0].set_title("Age by Passenger Class")
axes[0].set_xlabel("Passenger Class")
axes[0].set_ylabel("Age")

sns.boxplot(data=titanic, x="Pclass", y="Fare", ax=axes[1])
axes[1].set_title("Fare by Passenger Class")
axes[1].set_xlabel("Passenger Class")
axes[1].set_ylabel("Fare")

survival_by_class = titanic.groupby("Pclass")["Survived"].mean()
survival_by_class.plot(kind="bar", ax=axes[2])
axes[2].set_title("Survival Rate by Class")
axes[2].set_xlabel("Passenger Class")
axes[2].set_ylabel("Survival Rate")
axes[2].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()

print("\nMy note:")
print("First-class passengers were older on average and paid much higher fares.")
print("Their survival rate was also clearly higher than in third class.")
```

```python
# 3. Avocado Prices

url = "https://raw.githubusercontent.com/hosseinayoubi/EDA/main/avocado.csv"
avocado = pd.read_csv(url)

print("Shape:", avocado.shape)
print("Total missing values:", avocado.isna().sum().sum())

display(avocado.head())
display(avocado[["AveragePrice", "Total Volume", "year"]].describe().round(2))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(data=avocado, x="type", y="AveragePrice", ax=axes[0])
axes[0].set_title("Average Price by Type")
axes[0].set_xlabel("Type")
axes[0].set_ylabel("Average Price")

sns.boxplot(data=avocado, x="year", y="AveragePrice", ax=axes[1])
axes[1].set_title("Average Price by Year")
axes[1].set_xlabel("Year")
axes[1].set_ylabel("Average Price")

avocado["Date"] = pd.to_datetime(avocado["Date"])
total_us = avocado[avocado["region"] == "TotalUS"]

sns.lineplot(
    data=total_us,
    x="Date",
    y="Total Volume",
    hue="type",
    ax=axes[2]
)
axes[2].set_title("US Avocado Volume Over Time")
axes[2].set_xlabel("Date")
axes[2].set_ylabel("Total Volume")

plt.tight_layout()
plt.show()

print("\nMy note:")
print("Organic avocados are usually more expensive than conventional ones.")
print("2017 also looks more expensive, while conventional volume is much higher.")
```

```python
# 4. 80 Cereals Nutrition

url = "https://raw.githubusercontent.com/hosseinayoubi/EDA/main/cereal.csv"
cereal = pd.read_csv(url)

print("Shape:", cereal.shape)
print("Total missing values:", cereal.isna().sum().sum())

display(cereal.head())
display(cereal[["calories", "fiber", "sugars", "rating"]].describe().round(2))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.boxplot(data=cereal, x="mfr", y="sugars", ax=axes[0])
axes[0].set_title("Sugar by Manufacturer")
axes[0].set_xlabel("Manufacturer")
axes[0].set_ylabel("Sugar")

sns.boxplot(data=cereal, x="mfr", y="rating", ax=axes[1])
axes[1].set_title("Rating by Manufacturer")
axes[1].set_xlabel("Manufacturer")
axes[1].set_ylabel("Rating")

plt.tight_layout()
plt.show()

print("\nMy note:")
print("Sugar levels vary quite a lot between manufacturers.")
print("Ratings also differ, and manufacturer N looks strong in this small dataset.")
```

```python
# 5. Red Wine Quality

url = "https://raw.githubusercontent.com/hosseinayoubi/EDA/main/winequality-red.csv"
wine = pd.read_csv(url)

print("Shape:", wine.shape)
print("Total missing values:", wine.isna().sum().sum())

display(wine.head())
display(wine.describe().round(2))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.boxplot(data=wine, x="quality", y="alcohol", ax=axes[0])
axes[0].set_title("Alcohol by Wine Quality")
axes[0].set_xlabel("Quality")
axes[0].set_ylabel("Alcohol")

sns.boxplot(data=wine, x="quality", y="volatile acidity", ax=axes[1])
axes[1].set_title("Volatile Acidity by Wine Quality")
axes[1].set_xlabel("Quality")
axes[1].set_ylabel("Volatile Acidity")

plt.tight_layout()
plt.show()

print("\nMy note:")
print("Higher-rated wines generally have more alcohol.")
print("They also tend to have lower volatile acidity.")
```
